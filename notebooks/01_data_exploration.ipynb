# Rananiti Data Exploration

Inspect raw ball-by-ball and match coverage, column types, missingness, and basic cricket data distribution.

All inputs are loaded from `data/raw/`. Outputs are saved to `outputs/`.

```python
from pathlib import Path
import json, os, sys
import numpy as np
import pandas as pd
os.environ.setdefault("LOKY_MAX_CPU_COUNT", str(min(8, os.cpu_count() or 2)))
ROOT_DIR=Path.cwd()
if not (ROOT_DIR/"data"/"raw"/"ball_by_ball.csv").exists(): ROOT_DIR=ROOT_DIR.parent
sys.path.insert(0,str(ROOT_DIR))
from config import BALLS_FILE, MATCHES_FILE, OUTPUTS
OUTPUTS.mkdir(parents=True,exist_ok=True)
balls=pd.read_csv(BALLS_FILE,low_memory=False)
matches=pd.read_csv(MATCHES_FILE,low_memory=False) if MATCHES_FILE.exists() else pd.DataFrame()
print("Project:",ROOT_DIR,"| ball rows:",len(balls),"| matches:",matches.match_id.nunique() if not matches.empty else "metadata missing")
print("Ball data shape:",balls.shape)
print("Match metadata shape:",matches.shape)
print(); print("Ball-by-ball schema")
display(pd.DataFrame({"column":balls.columns,"dtype":balls.dtypes.astype(str).values,"missing":balls.isna().sum().values,"unique":balls.nunique(dropna=True).values}))
if "delivery_ist_time" in balls:
    dates=pd.to_datetime(balls.delivery_ist_time,errors="coerce",utc=True)
    print(); print("Delivery date range:",dates.min(),"to",dates.max())
print(); print("Matches by innings")
display(balls.groupby("innings").agg(deliveries=("match_id","size"),matches=("match_id","nunique")))
```
