# Rananiti Match State Reconstruction

Reconstruct score, wickets, balls remaining, chase target, and run rates before each delivery.

All inputs are loaded from `data/raw/`. Outputs are saved to `outputs/`.

```python
from pathlib import Path
import json, os, sys
import numpy as np
import pandas as pd
os.environ.setdefault("LOKY_MAX_CPU_COUNT", str(min(8, os.cpu_count() or 2)))
ROOT_DIR=Path.cwd()
if not (ROOT_DIR/"data"/"raw"/"ball_by_ball.csv").exists(): ROOT_DIR=ROOT_DIR.parent
sys.path.insert(0,str(ROOT_DIR))
from config import BALLS_FILE, MATCHES_FILE, OUTPUTS
OUTPUTS.mkdir(parents=True,exist_ok=True)
balls=pd.read_csv(BALLS_FILE,low_memory=False)
matches=pd.read_csv(MATCHES_FILE,low_memory=False) if MATCHES_FILE.exists() else pd.DataFrame()
print("Project:",ROOT_DIR,"| ball rows:",len(balls),"| matches:",matches.match_id.nunique() if not matches.empty else "metadata missing")
from src.pressure_engine.match_state import reconstruct_match_state
state=reconstruct_match_state(balls,matches if not matches.empty else None)
print("Reconstructed deliveries:",len(state))
cols=[c for c in ["match_id","innings","overid","over_ball_num","score_before","wickets_before","legal_balls_before","balls_remaining","wickets_remaining","current_run_rate","target","runs_required","required_run_rate","match_stage"] if c in state]
display(state[cols].head(20))
print(); print("Match-state coverage")
display(state.groupby("innings").agg(deliveries=("match_id","size"),matches=("match_id","nunique"),mean_score_before=("score_before","mean"),mean_wickets_before=("wickets_before","mean")))
```
